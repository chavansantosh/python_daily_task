# Take a user's first name, last name, and age. Print a single formatted greeting sentence using the .format() string method.

```python
first_name = input()
last_name = input()
age = int(input())

# Solution using .format()
print("Hello, {} {}! You are {} years old.".format(first_name, last_name, age))
```

#  Take a single word from the user and print it completely reversed.

```python
word = input()

# Solution using string slicing [start:stop:step]
reversed_word = word[::-1]
print(reversed_word)
```

#  Take an employee's full name (separated by a space). Convert it completely to lowercase and replace the space with a dot (.) to generate a company email address ending with @company.com.

```python
full_name = input()

# Solution using lower() and replace()
clean_name = full_name.lower().replace(" ", ".")
email = clean_name + "@company.com"
print(email)
```

#  Ask the user for their total shopping bill amount. If the bill is $100 or more, apply a flat 10% discount. Otherwise, no discount is given. Print the final bill amount rounded to 2 decimal places using f-strings.

```python
bill = float(input())

if bill >= 100:
    bill = bill * 0.90  # Deducting 10%

print(f"Final bill: ${bill:.2f}")
```

# Calculate a movie ticket price based on age rules: Under 5 years old is Free ($0), ages 5 to 18 pays $10, and anyone older pays $15.

```python
age = int(input())

if age < 5:
    price = 0
elif age <= 18:
    price = 10
else:
    price = 15

print(f"Ticket price is ${price}")
```

# Take an integer N from the user. Calculate and print the sum of all even numbers starting from 1 up to and including N.

```python
n = int(input())
total_sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        total_sum += i

print(total_sum)
```

# Write a program that keeps asking the user to type a password using a while loop. The correct password is "secret123". Give the user a maximum of 3 attempts before printing a lockout message.

```python
correct_password = "secret123"
attempts = 0

while attempts < 3:
    guess = input("Enter password: ")
    if guess == correct_password:
        print("Access Granted")
        break
    else:
        attempts += 1
        print("Incorrect password.")

if attempts == 3:
    print("Account Locked")
```

# Given a list of student marks, write a program that loops through the list and creates a new list containing only the passing marks (marks that are 40 or higher).

```python
marks = [35, 55, 80, 22, 41]
passing_marks = []

for score in marks:
    if score >= 40:
        passing_marks.append(score)

print(passing_marks)
```

#  Given a list of numbers, find the largest number without using Python's built-in max() function.

```python
numbers = [14, 52, 9, 31, 44]
largest = numbers[0]  # Start by assuming the first number is the biggest

for num in numbers:
    if num > largest:
        largest = num

print(largest)
```
