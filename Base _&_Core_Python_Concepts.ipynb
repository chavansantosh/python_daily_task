# 1. Reverse a string without using slicing or built-in reverse functions.

```python
def reverse_string(s):
    reversed_str = ""
    for char in s:
        reversed_str = char + reversed_str
    return reversed_str

print(reverse_string("hello"))  # Output: olleh
```

# 2. Check if a string is a palindrome.

```python
def is_palindrome(s):
    cleaned = "".join(char.lower() for char in s if char.isalnum())
    return cleaned == cleaned[::-1]

print(is_palindrome("A man, a plan, a canal: Panama"))  
```

# 3. Find the largest and smallest numbers in a list without using max() or min()

```python
def find_min_max(nums):
    if not nums:
        return None, None
    smallest = largest = nums[0]
    for num in nums[1:]:
        if num < smallest:
            smallest = num
        if num > largest:
            largest = num
    return smallest, largest

print(find_min_max([4, 1, 9, 7, -3, 5]))  
```

# 4. Count the frequency of each character in a string.

```python
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

print(char_frequency("banana"))  
```

# 5. Remove duplicates from a list while preserving the original order.

```python
def remove_duplicates(nums):
    return list(dict.fromkeys(nums))

print(remove_duplicates([1, 2, 2, 3, 1, 4, 2])) 
```

# 6. Check if two strings are anagrams of each other.

```python
from collections import Counter

def are_anagrams(s1, s2):
    return Counter(s1.replace(" ", "").lower()) == Counter(s2.replace(" ", "").lower())

print(are_anagrams("Listen", "Silent"))  
```

# 7. Find the factorial of a number using recursion.

```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if n in (0, 1):
        return 1
    return n * factorial(n - 1)

print(factorial(5))  
```

# 8. Generate the Fibonacci sequence up to $n$ terms. 

```python
def generate_fibonacci(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]
    fib = [0, 1]
    for _ in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib

print(generate_fibonacci(7))  
```

# 9. Check if a number is prime.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(11))  
```

# 10. Flatten a nested list (arbitrary depth).

```python
def flatten_list(nested):
    flat = []
    for item in nested:
        if isinstance(item, list):
            flat.extend(flatten_list(item))
        else:
            flat.append(item)
    return flat

print(flatten_list([1, [2, [3, 4], 5], 6]))  
```

# 11. Find the second largest element in a list.

```python
def second_largest(nums):
    unique_nums = set(nums)
    if len(unique_nums) < 2:
        return None
    unique_nums.remove(max(unique_nums))
    return max(unique_nums)

print(second_largest([10, 20, 4, 45, 99, 99]))  
```

# 12. Merge two sorted lists into a single sorted list.

```python
def merge_sorted_lists(l1, l2):
    i = j = 0
    merged = []
    while i < len(l1) and j < len(l2):
        if l1[i] < l2[j]:
            merged.append(l1[i])
            i += 1
        else:
            merged.append(l2[j])
            j += 1
    merged.extend(l1[i:])
    merged.extend(l2[j:])
    return merged

print(merge_sorted_lists([1, 3, 5], [2, 4, 6, 8]))  
```

## 13. Find the missing number in an array containing numbers from $1$ to $n$.

```python
def find_missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(find_missing_number([1, 2, 4, 5, 6], 6))  # Output: 3
```

# 14. Find all pairs in an integer array that sum up to a specific target (Two Sum).

```python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        diff = target - num
        if diff in seen:
            return [seen[diff], i]
        seen[num] = i
    return []

print(two_sum([2, 7, 11, 15], 9)) 
```

# 15. Rotate a list to the left by $k$ positions.

```python
def rotate_left(nums, k):
    k = k % len(nums)
    return nums[k:] + nums[:k]

print(rotate_left([1, 2, 3, 4, 5], 2))  
```

# 16. Implement Binary Search on a sorted list

```python
def binary_search(nums, target):
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

print(binary_search([2, 5, 8, 12, 16, 23, 38], 16))  
```

# 17. Find the intersection of two lists.

```python
def list_intersection(l1, l2):
    return list(set(l1) & set(l2))

print(list_intersection([1, 2, 3, 4], [3, 4, 5, 6]))  
```

# 18. Check if a string has all unique characters.

```python
def has_unique_chars(s):
    return len(set(s)) == len(s)

print(has_unique_chars("python"))  
print(has_unique_chars("hello"))   
```

# 19. Count the number of vowels in a string.

```python
def count_vowels(s):
    vowels = set("aeiouAEIOU")
    return sum(1 for char in s if char in vowels)

print(count_vowels("Hello World")) 
```

# 20. Find the longest word in a sentence.

```python
def longest_word(sentence):
    words = sentence.split()
    if not words:
        return ""
    return max(words, key=len)

print(longest_word("Python is an awesome programming language")) 
```
