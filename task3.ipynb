# 1. How do you reverse a string without using built-in functions?
# Answer: Use sequence slicing [::-1] or iterate backward using a loop.

```python
def reverse_string(s: str) -> str:
    reversed_str = ""
    for char in s:
        reversed_str = char + reversed_str
    return reversed_str

text = "Python"
print(reverse_string(text))
print(text[::-1])
```

# 2. What is the difference between is and ==?
# Answer: == checks for value equality, while is checks for object identity (i.e., whether two variables refer to the exact same memory address).

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)
print(a is b)
print(a is c)
```

# 3. How do decorators work in Python?
# Answer: A decorator is a function that takes another function as an argument and extends its behavior without modifying it explicitly.

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

# 5. What are list comprehensions and generator expressions?
# Answer: List comprehensions create lists in memory immediately using [], whereas generator expressions produce items one by one lazily using ().

```python
squared_list = [x**2 for x in range(5)]
print(squared_list)

squared_gen = (x**2 for x in range(5))
print(list(squared_gen))
```

# 6. How do you find non-duplicate (unique) elements while preserving order?
# Answer: Use dict.fromkeys() or loop through maintaining a set for tracked items.

```python
def remove_duplicates(items: list) -> list:
    return list(dict.fromkeys(items))

data = [1, 3, 2, 1, 4, 3, 5]
print(remove_duplicates(data))
```

# 7. How do *args and **kwargs work in function arguments?
# Answer: *args collects positional arguments into a tuple, while **kwargs collects keyword arguments into a dictionary.

```python
def show_args(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)

show_args(1, 2, 3, name="Alice", role="Dev")
```

# 8. How do you handle exceptions using try-except-else-finally?
# Answer: try runs code that might fail, except catches errors, else runs if no error occurs, and finally runs regardless.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero!")
    else:
        print(f"Result: {result}")
    finally:
        print("Execution completed.")

safe_divide(10, 2)
safe_divide(10, 0)
```

# 9. How do you flattening a nested list in Python?
# Answer: Use a recursive function or list comprehension.

```python
def flatten(nested_list):
    flat = []
    for item in nested_list:
        if isinstance(item, list):
            flat.extend(flatten(item))
        else:
            flat.append(item)
    return flat

nested = [1, [2, [3, 4], 5], 6]
print(flatten(nested))
```

# 10. What is the difference between shallow copy and deep copy?
# Answer: Shallow copies duplicate reference structures; deep copies recursively copy referenced nested objects.

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0][0] = 99

print("Shallow:", shallow)
print("Deep:", deep)
```

# 11. Find the first non-repeating character in a string.

```python
from collections import Counter

def first_non_repeating_char(s):
    counts = Counter(s)
    for char in s:
        if counts[char] == 1:
            return char
    return None

print(first_non_repeating_char("swiss"))  
```

# 12.Write a Python program to print the first n rows of Pascal’s Triangle.
Use nested loops to calculate combination values.
Each number is the sum of two numbers above it.
Print 1 at the beginning and end of each row.
Display the triangle in a structured format.


```python
n = int(input())

for i in range(n):
  row = []
  val = 1
  for j in range(i + 1):
    row.append(str(val))
    # Calculate the next value in the row using combination logic
    val = val * (i - j) // (j + 1)
  print(" ".join(row))
```

# 13.Write a Python program to print the letter Z using *.
# Print a full row of stars at the top.
# Print diagonal stars using spaces and a single *.
# Print another full row of stars at the bottom.
# Use nested loops to manage alignment.

```python
n = int(input().strip())

for i in range(n):
  for j in range()
    if i == 0 or i == n - 1 or j == n - 1 - i:
      print("*", end="")
    else:
      print(" ", end="")
  print()
```
