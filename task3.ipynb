# 1. How do you reverse a string without using built-in functions?
# Answer: Use sequence slicing [::-1] or iterate backward using a loop.

```python
def reverse_string(s: str) -> str:
    reversed_str = ""
    for char in s:
        reversed_str = char + reversed_str
    return reversed_str

text = "Python"
print(reverse_string(text))
print(text[::-1])
```

# 2. What is the difference between is and ==?
# Answer: == checks for value equality, while is checks for object identity (i.e., whether two variables refer to the exact same memory address).

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)
print(a is b)
print(a is c)
```

# 3. How do decorators work in Python?
# Answer: A decorator is a function that takes another function as an argument and extends its behavior without modifying it explicitly.

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

# 5. What are list comprehensions and generator expressions?
# Answer: List comprehensions create lists in memory immediately using [], whereas generator expressions produce items one by one lazily using ().

```python
squared_list = [x**2 for x in range(5)]
print(squared_list)

squared_gen = (x**2 for x in range(5))
print(list(squared_gen))
```

# 8. How do you handle exceptions using try-except-else-finally?
# Answer: try runs code that might fail, except catches errors, else runs if no error occurs, and finally runs regardless.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero!")
    else:
        print(f"Result: {result}")
    finally:
        print("Execution completed.")

safe_divide(10, 2)
safe_divide(10, 0)
```

# 9. How do you flattening a nested list in Python?
# Answer: Use a recursive function or list comprehension.

```python
def flatten(nested_list):
    flat = []
    for item in nested_list:
        if isinstance(item, list):
            flat.extend(flatten(item))
        else:
            flat.append(item)
    return flat

nested = [1, [2, [3, 4], 5], 6]
print(flatten(nested))
```

# 10. What is the difference between shallow copy and deep copy?
# Answer: Shallow copies duplicate reference structures; deep copies recursively copy referenced nested objects.

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0][0] = 99

print("Shallow:", shallow)
print("Deep:", deep)
```

# 11. Find the first non-repeating character in a string.

```python
from collections import Counter

def first_non_repeating_char(s):
    counts = Counter(s)
    for char in s:
        if counts[char] == 1:
            return char
    return None

print(first_non_repeating_char("swiss"))  
```

# 12.Write a Python program to print the first n rows of Pascal’s Triangle.
Use nested loops to calculate combination values.
Each number is the sum of two numbers above it.
Print 1 at the beginning and end of each row.
Display the triangle in a structured format.


```python
n = int(input())

for i in range(n):
  row = []
  val = 1
  for j in range(i + 1):
    row.append(str(val))
    # Calculate the next value in the row using combination logic
    val = val * (i - j) // (j + 1)
  print(" ".join(row))
```

# 13.Write a Python program to print the letter Z using *.
# Print a full row of stars at the top.
# Print diagonal stars using spaces and a single *.
# Print another full row of stars at the bottom.
# Use nested loops to manage alignment.

```python
n = int(input().strip())

for i in range(n):
  for j in range()
    if i == 0 or i == n - 1 or j == n - 1 - i:
      print("*", end="")
    else:
      print(" ", end="")
  print()
```

# 14.Write a Python program to print a butterfly pattern using *.
# Use two halves: an increasing pattern and a decreasing pattern.
# Print stars on both left and right sides of each row.
# Add spaces between the star groups.
# Ensure symmetry on both sides.

```python
n = int(input().strip())
for i in range(1, n + 1):
  print("*" * i + " " * (2 * (n - i)) + "*" * i)
for i in range(n, 0, -1):
  print("*" * i + " " * (2 * (n - i)) + "*" * i)
```

# 15 Write a program that swaps keys and values in a dictionary
#  The user will first enter how many key-value pairs they want to add
#  Then they'll enter each key and its corresponding value
#  Create a new dictionary where the original values become keys and original keys become values
#  Display the new swapped dictionary
#  This teaches you how to transform dictionaries by reversing the relationship between keys and values

```python
n = int(input())
data = {}
for _ in range(n):
    key = input().strip()
    value = input().strip()
    data[key] = value
swapped_dict = {val: k for k, val in data.items()}
print(swapped_dict)
```

# 16.Write a Python program to count how many times each character appears in a string.
# The user enters a string via console.
# Store counts in a dictionary with characters as keys and counts as values.

```python
s = input()

char_counts = {}

for char in s:
    char_counts[char] = char_counts.get(char, 0) + 1
print(char_counts)
```

# 17.Write a Python program to combine two dictionaries.
# The user enters both dictionaries via console.
# If a key exists in both dictionaries, their numeric values are added.

```python
n1 = int(input())
dict1 = {}

for i in range(n1):
    key = input()
    value = int(input())  # or float(input()) depending on test cases
    dict1[key] = value
n2 = int(input())
dict2 = {}

for i in range(n2):
    key = input()
    value = int(input())
    dict2[key] = value

# Merge dictionaries and add values for common keys
result_dict = dict1.copy()

for key, value in dict2.items():
    if key in result_dict:
        result_dict[key] += value
    else:
        result_dict[key] = value
print(result_dict)
```

# 18.Write a Python program to sort a dictionary by its values in descending order.
# The user enters the dictionary via console.
# Print the sorted key-value pairs as a list of tuples.

```python
n = int(input("Enter number of items in dictionary: "))
d = {}

for i in range(1, n + 1):
    key = input(f"Enter key {i}: ")
    val_str = input(f"Enter value {i}: ")
    
    # Convert value to int or float depending on input format
    if '.' in val_str:
        val = float(val_str)
    else:
        val = int(val_str)
        
    d[key] = val

# Sort the dictionary items by value (index 1 of key-value tuple) in descending order
sorted_items = sorted(d.items(), key=lambda item: item[1], reverse=True)

print(sorted_items)
```

# 19.Write a Python program to check whether one set is a subset or superset of another.
# The program should read two sets from the console.
# Use the issubset() method to verify if the first set is contained in the second.
# Use the issuperset() method to check if the first set contains all elements of the second.
# Print the results clearly.

```python
line1 = input().strip()
line2 = input().strip()

set_a = set(line1.split())
set_b = set(line2.split())

# Print the results as expected by the test cases
print(set_a.issubset(set_b))
print(set_b.issuperset(set_a))
```
