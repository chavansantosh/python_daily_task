# 1.This program demonstrates method overriding in Python.
# Create a parent class Employee with a method show_role().
# Create a child class Manager that overrides the show_role() method.
# The overridden method prints the employee name followed by Manager.


```python
class Employee:
    def __init__(self, name):
        self.name = name

    def show_role(self):
        print(f"{self.name} - Employee")


class Manager(Employee):
    def show_role(self):
        print(f"{self.name} - Manager")


# Read input from standard input
if __name__ == "__main__":
    name = input().strip()
    # Create an instance of the Manager child class
    mgr = Manager(name)
    # Call the overridden show_role method
    mgr.show_role()
```

# 2.This program demonstrates how a child class can extend the behavior of a parent class using inheritance.
# Create a parent class Employee with a method calculate_salary(basic) that returns the basic salary.
# Create a child class FullTimeEmployee that calls the parent method and adds a 20% bonus.

```python
class Employee:
    def calculate_salary(self, basic):
        return basic


class FullTimeEmployee(Employee):
    def calculate_salary(self, basic):
        # Call the parent class method and add a 20% bonus
        base_salary = super().calculate_salary(basic)
        return int(base_salary * 1.20)


# Read input from standard input
if __name__ == "__main__":
    basic = int(input().strip())
    # Create an instance of FullTimeEmployee
    emp = FullTimeEmployee()
    # Calculate and print the final salary
    final_salary = emp.calculate_salary(basic)
    print(final_salary)
```

# 5.Create a class PerformanceScore with method get_score(). 
# Create another class Experience with method get_experience(). 
# Create a child class Appraisal that gives hike if score > 80 and experience > 3 years.

```python
class PerformanceScore:
    def __init__(self, score):
        self.score = score

    def get_score(self):
        return self.score


class Experience:
    def __init__(self, experience):
        self.experience = experience

    def get_experience(self):
        return self.experience


class Appraisal(PerformanceScore, Experience):
    def __init__(self, score, experience):
        PerformanceScore.__init__(self, score)
        Experience.__init__(self, experience)

    def check_appraisal(self):
        if self.get_score() > 80 and self.get_experience() > 3:
            return "Hike Approved"
        return "Hike Not Approved"


# Reading input and executing
if __name__ == "__main__":
    score, exp = map(int, input().split())
    emp = Appraisal(score, exp)
    print(emp.check_appraisal())
```

```python
# Base class
class Customer:
    def __init__(self, income):
        self.income = income

    def get_income(self):
        return self.income


# Intermediate child class inheriting from Customer
class LoanCheck(Customer):
    def calculate_emi(self):
      
        return 10000


# Derived child class inheriting from LoanCheck
class Approval(LoanCheck):
    def check_approval(self):
        income = self.get_income()
        emi = self.calculate_emi()

        if emi < 0.40 * income:
            return "Approved"
        else:
            return "Rejected"


if __name__ == "__main__":
    customer_income = float(input().strip())
    loan_application = Approval(customer_income)
    print(loan_application.check_approval())
```

# 7.Create a base class Employee with a method get_score() to return the employee score.
# Create a child class Performance that calculates rating (optional for internal use).
# Create another child class Appraisal that gives hike percentage based on score:
# 20% hike if score ≥ 80
# 10% hike if score < 80

```python
# Base class
class Employee:
    def __init__(self, score):
        self.score = score

    def get_score(self):
        return self.score


# Intermediate child class inheriting from Employee
class Performance(Employee):
    def calculate_rating(self):
        score = self.get_score()
        if score >= 80:
            return "Exceeds"
        elif score >80:
            return "Meets"
        else:
            return "Needs Improvement"


# Derived child class inheriting from Performance
class Appraisal(Performance):
    def get_hike(self):
        score = self.get_score()
        _ = self.calculate_rating()

        if score >= 80:
            return "20 percent Hike"
        elif score >= 50:
            return "10 percent Hike"
        else:
            return "No Hike"
            
if __name__ == "__main__":
    emp_score = float(input().strip())
    appraisal = Appraisal(emp_score)
    print(appraisal.get_hike())
```

# 8.Base class Parking: method get_hours() to get the number of parking hours.
# Child class FeeCalculator: calculates parking fee based on fixed hourly rate.
# Child class FinalFee: applies extra charge if hours exceed 5.

```python
class Parking:
    def __init__(self, hours: int):
        self.hours = hours

    def get_hours(self) -> int:
        return self.hours


class FeeCalculator(Parking):
    def __init__(self, hours: int, hourly_rate: int = 10):
        super().__init__(hours)
        self.hourly_rate = hourly_rate

    def calculate_base_fee(self) -> int:
        return self.get_hours() * self.hourly_rate


class FinalFee(FeeCalculator):
    def __init__(self, hours: int, hourly_rate: int = 10, extra_charge: int = 10):
        super().__init__(hours, hourly_rate)
        self.extra_charge = extra_charge

    def calculate_total_fee(self) -> int:
        fee = self.calculate_base_fee()
        if self.get_hours() > 5:
            fee += self.extra_charge
        return fee


if __name__ == "__main__":
    hours_input = int(input().strip())
    parking_fee = FinalFee(hours=hours_input)
    print(parking_fee.calculate_total_fee())
```

# 9.You need to create a program that calculates interest on different types of loans.
# There are two types of loans:

# HomeLoan: The interest rate is 8% per year on the loan amount.
# PersonalLoan: The interest rate is 12% per year on the loan amount.

# Input Format:

# First line: Loan type (either "HomeLoan" or "PersonalLoan")
# Second line: Loan amount (a number)

# Output Format:

# Print the interest amount (rounded to 2 decimal places)

```python
class Loan:
    def __init__(self, amount: float):
        self.amount = amount

    def calculate_interest(self) -> float:
        raise NotImplementedError("Subclasses must implement calculate_interest()")


class HomeLoan(Loan):
    def calculate_interest(self) -> float:
        return self.amount * 0.08


class PersonalLoan(Loan):
    def calculate_interest(self) -> float:
        return self.amount * 0.12


if __name__ == "__main__":
    loan_type = input().strip()
    loan_amount = float(input().strip())

    if loan_type == "HomeLoan":
        loan = HomeLoan(loan_amount)
    elif loan_type == "PersonalLoan":
        loan = PersonalLoan(loan_amount)
    else:
        raise ValueError("Invalid loan type")

    print(f"{loan.calculate_interest():.2f}")
```

# 11.You need to create a program that calculates the final bill amount for hospital patients based on their insurance status.

# 
# InsuredPatient: Gets a 30% discount on the total bill amount.
# UninsuredPatient: Pays the full bill amount without any discount.
#

```python
class Loan:
    def calculate_emi(self, amount):
        return amount


class HomeLoan(Loan):
    def __init__(self, years=10):
        self.years = years

    def calculate_emi(self, amount):
        # Overrides parent method to divide amount by number of years
        return amount // self.years 

amount = int(input())
home_loan = HomeLoan(years=10)
print(home_loan.calculate_emi(amount))
```

# 16.Create a class Performance with an attribute score.
# Overload the >= (greater than or equal to) operator using the __ge__() method to check whether an employee meets the promotion eligibility criteria.

```python
class Performance:
    def __init__(self, score):
        self.score = score

    def __ge__(self, other):
        # Supports comparing against another Performance object or a direct numeric threshold
        if isinstance(other, Performance):
            return self.score >= other.score
        return self.score >= other
        
emp_score, min_criteria = map(int, input().split())

employee = Performance(emp_score)
criteria = Performance(min_criteria)

if employee >= criteria:
    print("Eligible")
else:
    print("Not Eligible")
```

# 17.Create a class Cart with an attribute total.
# Overload the == (equality) operator using the __eq__() method to check whether two shopping carts have the same total value.

```python
class Cart:
    def __init__(self, total):
        self.total = total

    def __eq__(self, other):
        # Returns True if comparing two Cart objects with equal totals
        if isinstance(other, Cart):
            return self.total == other.total
        return False


# Example usage
total1, total2 = map(float, input().split())

cart1 = Cart(total1)
cart2 = Cart(total2)

if cart1 == cart2:
    print("Equal")
else:
    print("Not Equal")
```

# 18.Reads a Roman numeral string and converts it to its integer equivalent. Validates proper Roman numeral format before conversion.

```python
import sys
import re

def romanToInt(s: str) -> int:
    s = s.strip().upper()
    roman_pattern = r"^M{0,3}(CM|CD|D?C{0,3})(XC|XL|L?X{0,3})(IX|IV|V?I{0,3})$"
    
    if not s or not re.match(roman_pattern, s):
        raise ValueError(f"Invalid Roman numeral format: '{s}'")
    
    values = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    total = 0
    i = 0
    n = len(s)
    
    while i < n:
        curr_val = values[s[i]]
        if i + 1 < n and curr_val < values[s[i + 1]]:
            total += values[s[i + 1]] - curr_val
            i += 2
        else:
            total += curr_val
            i += 1
            
    return total

if __name__ == "__main__":
   
    for line in sys.stdin:
        test_case = line.strip()
        if test_case:
            print(romanToInt(test_case))
```

# 23.Write a recursive function to find the Nth Fibonacci number.
# The input is a single integer.
# Follow the Fibonacci sequence definition.
# The function should make recursive calls as required.
# Define base cases for the first two numbers.
# Do not use loops or lists.
# Return the Fibonacci value.
# Print the final output.

```python
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(int(input().strip())))
```

# 24.Write a recursive function to calculate the power of a number.
# The input consists of base and exponent in a single line.
# Multiply the base recursively until exponent becomes zero.
# Define a base condition for exponent zero.
# Do not use the ** operator.
# Return the calculated power value.
# Print the final result.
# Ensure correct handling of zero exponent.

```python
def power(base, exp):
    if exp == 0:
        return 1
    # Recursive step: multiply base by power(base, exp - 1)
    return base * power(base, exp - 1)

input_data = input().split()
base = int(input_data[0])
exponent = int(input_data[1])

# Calculate and print the final result
print(power(base, exponent))
```

# 25.Write a recursive function to reverse a given number.
# The input is a single integer.
# Process one digit at a time in each recursive call.
# Reduce the number size in every step.
# Do not convert the number to a string.
# Return the reversed number.
# Print the final output.
# Handle trailing zeros correctly.

```python
def reverse_number(n, rev=0):
    if n == 0:
        return rev
    digit = n % 10
    # Recursive call reducing the number size by integer division of 10
    return reverse_number(n // 10, rev * 10 + digit)

# Read the single integer input
input_number = int(input().strip())

# Compute and print the reversed number
print(reverse_number(input_number))
```
